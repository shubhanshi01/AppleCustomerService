```python
import re
import html
import unicodedata
```

```python
def clean_customer_text(text):
    """
    Conservative cleaning for AppleSupport customer tweets.

    Keeps:
    - technical terms
    - numbers
    - product names
    - error messages
    - hashtags as words

    Removes:
    - URLs
    - @mentions
    - emojis
    - unnecessary Unicode artifacts
    - excessive whitespace
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # ----------------------------------------
    # 1. Decode HTML entities
    # ----------------------------------------
    text = html.unescape(text)

    # ----------------------------------------
    # 2. Remove URLs
    # ----------------------------------------
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # ----------------------------------------
    # 3. Remove @mentions
    # ----------------------------------------
    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # ----------------------------------------
    # 4. Keep hashtag word, remove '#'
    # ----------------------------------------
    text = re.sub(
        r"#(\w+)",
        r"\1",
        text
    )

    # ----------------------------------------
    # 5. Unicode normalization
    # ----------------------------------------
    text = unicodedata.normalize(
        "NFKC",
        text
    )

    # ----------------------------------------
    # 6. Remove variation selectors
    #    and zero-width characters
    # ----------------------------------------
    text = re.sub(
        r"[\u200b-\u200f\u202a-\u202e\ufeff\ufe0e\ufe0f]",
        "",
        text
    )

    # ----------------------------------------
    # 7. Remove emojis / symbols
    #
    # Keep normal letters, numbers and
    # punctuation.
    # ----------------------------------------
    cleaned_chars = []

    for char in text:
        category = unicodedata.category(char)

        if category.startswith("So"):
            continue

        if category.startswith("Sk"):
            continue

        cleaned_chars.append(char)

    text = "".join(cleaned_chars)

    # ----------------------------------------
    # 8. Normalize repeated whitespace
    # ----------------------------------------
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # ----------------------------------------
    # 9. Strip whitespace
    # ----------------------------------------
    text = text.strip()

    return text
```

```python
def initial_quality_flag(text):
    """
    Initial conservative quality filtering.
    Does not permanently delete anything.
    """

    if not text or not text.strip():
        return "EMPTY_AFTER_CLEANING"

    text_lower = text.lower().strip()

    # Very short acknowledgement / conversational messages
    low_information = {
        "thanks",
        "thank you",
        "thx",
        "ok",
        "okay",
        "yes",
        "no",
        "yep",
        "nope",
        "sure",
        "i will",
        "will do",
        "great",
        "worked",
        "it worked"
    }

    if text_lower in low_information:
        return "LOW_INFORMATION"

    # Routing-only messages
    routing_only_patterns = [
        r"^dm sent$",
        r"^sent (you )?a dm$",
        r"^sent you (all )?my dm$",
        r"^i('?ve| have) sent (you )?(a )?dm$",
        r"^please check (your )?dms?$",
        r"^check (your )?dms?$",
        r"^dm me( please)?$",
        r"^please dm me$",
    ]

    for pattern in routing_only_patterns:
        if re.fullmatch(pattern, text_lower):
            return "ROUTING_ONLY"

    return "KEEP"
```

```python
apple_customer_tweets["quality_flag"] = (
    apple_customer_tweets["text_clean"]
    .apply(initial_quality_flag)
)
```

```python
def detect_language(text):
    """
    Conservative language detection.

    Very short / technical messages are marked unknown
    because language detection is unreliable for them.
    """

    if pd.isna(text):
        return "unknown"

    text = str(text).strip()

    # Too short for reliable language detection
    if len(text) < 20:
        return "unknown"

    # Need enough alphabetic characters
    alpha_chars = sum(c.isalpha() for c in text)

    if alpha_chars < 10:
        return "unknown"

    try:
        return detect(text)
    except:
        return "unknown"
```

```python
apple_customer_tweets["language"] = (
    apple_customer_tweets["text_clean"]
    .apply(detect_language)
)
```
